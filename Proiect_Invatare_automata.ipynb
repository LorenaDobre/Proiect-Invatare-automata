Importarea bibliotecilor

```python

# Importam biblioteca Pandas pentru citirea si analiza datelor
import pandas as pd

# Importam NumPy pentru operatii si calcule numerice
import numpy as np

# Importam Matplotlib pentru realizarea graficelor
import matplotlib.pyplot as plt

# Importam Seaborn pentru vizualizari statistice
import seaborn as sns

# Setam stilul graficelor pentru o afisare mai clara
sns.set_theme(style="whitegrid")

# Permitem afisarea tuturor coloanelor din dataset
pd.set_option("display.max_columns", None)

# Verificam daca bibliotecile au fost importate corect
print("Bibliotecile au fost importate cu succes!")
```

Încărcarea datasetului

```python

# Stabilim adresa de unde citim datasetul Telco Customer Churn
url = "https://raw.githubusercontent.com/rstudio/keras-customer-churn/master/data/WA_Fn-UseC_-Telco-Customer-Churn.csv"

# Citim fisierul CSV si salvam datele intr-un DataFrame numit df
df = pd.read_csv(url)

# Afisam un mesaj pentru a confirma incarcarea datelor
print("Datasetul a fost incarcat cu succes!")
```

Explorarea datasetului. Afișăm dimensiunea, primele rânduri și numele coloanelor.

```python

# Afisam dimensiunea datasetului
# shape returneaza (numar_randuri, numar_coloane)
print("Dimensiunea datasetului:")
print(df.shape)

# Afisam primele 5 randuri pentru a vedea cum sunt organizate datele
print("\nPrimele 5 randuri:")
display(df.head())

# Afisam numele tuturor coloanelor din dataset
print("\nNumele coloanelor:")
print(df.columns.tolist())
```

Verificăm tipul fiecărei coloane și valorile disponibile

```python

# Afisam informatii generale despre dataset
# info() arata numarul de valori nenule si tipul fiecarei coloane
print("Informatii generale despre dataset:")
df.info()

# Afisam separat tipul de date pentru fiecare coloana
# De exemplu: int64 pentru numere intregi, float64 pentru numere reale
# si object pentru coloane care contin de obicei text
print("\nTipurile coloanelor:")
display(df.dtypes)
```

Statistici descriptive. Calculăm media, minimul, maximul și numărul valorilor distincte.

```python

# Calculam statistici descriptive pentru coloanele numerice
# describe() ofera media, minimul, maximul, quartilele etc.
# .T transpune tabelul pentru a fi mai usor de citit
print("Statistici descriptive pentru coloanele numerice:")
display(df.describe().T)

# Calculam numarul de valori distincte din fiecare coloana
# nunique() numara valorile diferite, fara valorile lipsa
print("\nNumarul valorilor distincte pentru fiecare coloana:")
display(df.nunique())
```

Identificăm valorile lipsă, duplicatele și valorile goale

```python

# Verificam cate valori lipsa exista in fiecare coloana
# isnull() identifica valorile lipsa, iar sum() le numara
print("Valori lipsa pentru fiecare coloana:")
display(df.isnull().sum())

# Verificam daca exista randuri complet identice in dataset
# duplicated() identifica duplicatele, iar sum() le numara
print("\nNumarul randurilor duplicate:")
print(df.duplicated().sum())

# Verificam daca exista clienti cu acelasi identificator
# customerID ar trebui sa fie unic pentru fiecare client
print("\nNumarul ID-urilor duplicate:")
print(df["customerID"].duplicated().sum())

# Verificam daca TotalCharges contine valori goale sau doar spatii
# astype(str) transforma valorile in text
# str.strip() elimina spatiile de la inceput si de la sfarsit
# Comparatia cu "" identifica valorile ramase goale
print("\nValori goale in coloana TotalCharges:")
print((df["TotalCharges"].astype(str).str.strip() == "").sum())
```

Conversia TotalCharges. Transformăm costul total în format numeric

```python

# Cream o copie a datasetului original pentru analiza exploratorie
# Astfel, pastram datele initiale nemodificate in variabila df
df_eda = df.copy()

# Transformam coloana TotalCharges din text in valori numerice
# errors="coerce" transforma valorile invalide in NaN (valori lipsa)
df_eda["TotalCharges"] = pd.to_numeric(
    df_eda["TotalCharges"],
    errors="coerce"
)

# Verificam tipul de date dupa conversie
# Ne asteptam sa obtinem un tip numeric, de exemplu float64
print("Tipul coloanei TotalCharges:")
print(df_eda["TotalCharges"].dtype)

# Verificam cate valori lipsa exista dupa conversie
# Aceste valori vor fi tratate ulterior, in etapa de preprocesare
print("\nValori lipsa dupa conversie:")
print(df_eda["TotalCharges"].isnull().sum())
```

Distribuția Churn. Calculăm numărul și procentul clienților care au plecat sau au rămas

```python

# Afisam numarul de clienti pentru fiecare valoare a variabilei Churn
# No = clientul nu a renuntat la serviciile companiei
# Yes = clientul a renuntat la serviciile companiei
print("Distributia variabilei Churn:")
print(df_eda["Churn"].value_counts())

# Calculam procentul clientilor din fiecare categorie
# normalize=True calculeaza proportiile in locul numarului de clienti
# Inmultim cu 100 pentru a transforma proportiile in procente
# round(2) rotunjeste rezultatele la doua zecimale
print("\nDistributia procentuala:")
print(df_eda["Churn"].value_counts(normalize=True) * 100)
```

Reprezentăm grafic distribuția clienților

```python

# Stabilim dimensiunea graficului
plt.figure(figsize=(8, 5))

# Construim un grafic cu numarul clientilor din fiecare categorie
# x reprezinta variabila analizata
# order stabileste ordinea de afisare a categoriilor
sns.countplot(
    data=df_eda,
    x="Churn",
    order=["No", "Yes"],
    color="steelblue"
)

# Adaugam titlul graficului
plt.title("Distributia clientilor in functie de Churn")

# Denumim axele graficului
plt.xlabel("A renuntat clientul la servicii?")
plt.ylabel("Numarul de clienti")

# Afisam graficul
plt.show()
```

Analiza contractelor. Calculăm rata abandonului pentru fiecare tip de contract.

```python

# Calculam numarul de clienti pentru fiecare combinatie Contract-Churn
# crosstab() creeaza un tabel de frecvente pentru doua variabile
contract_counts = pd.crosstab(
    df_eda["Contract"],
    df_eda["Churn"]
)

# Afisam tabelul cu numarul de clienti
print("Numarul clientilor in functie de contract si Churn:")
display(contract_counts)

# Calculam distributia procentuala pentru fiecare tip de contract
# normalize="index" calculeaza procentele separat pentru fiecare rand
# Inmultim cu 100 pentru a obtine procente
contract_percent = pd.crosstab(
    df_eda["Contract"],
    df_eda["Churn"],
    normalize="index"
) * 100

# Afisam procentele rotunjite la doua zecimale
print("\nDistributia procentuala in functie de contract:")
display(contract_percent.round(2))
```

Comparăm grafic ratele abandonului în funcție de contract

```python

# Selectam procentul clientilor care au renuntat la servicii (Yes)
# Valorile provin din tabelul procentual calculat anterior
churn_contract = contract_percent["Yes"]

# Stabilim dimensiunea graficului
plt.figure(figsize=(9, 5))

# Construim un grafic cu bare pentru rata abandonului
# Fiecare bara reprezinta procentul clientilor care au plecat
ax = sns.barplot(
    x=churn_contract.index,
    y=churn_contract.values,
    color="steelblue"
)

# Afisam procentul deasupra fiecarei bare
for i, valoare in enumerate(churn_contract.values):
    ax.text(
        i, valoare + 1,
        f"{valoare:.2f}%",
        ha="center"
    )

# Adaugam titlul graficului
plt.title("Rata abandonului in functie de tipul contractului")

# Denumim axele graficului
plt.xlabel("Tipul contractului")
plt.ylabel("Rata abandonului (%)")

# Stabilim intervalul axei Y, astfel incat procentele sa fie vizibile
plt.ylim(0, 100)

# Afisam graficul
plt.show()
```

Calculăm statistici despre vechimea clienților (tenure)

```python

# Afisam statistici generale pentru vechimea clientilor
print("Statistici generale pentru tenure:")
display(df_eda["tenure"].describe())

# Grupam clientii dupa variabila Churn
# Pentru fiecare categorie calculam numarul, media si mediana vechimii
tenure_stats = df_eda.groupby("Churn")["tenure"].agg(
    ["count", "mean", "median", "min", "max"]
)

# Rotunjim rezultatele la doua zecimale
tenure_stats = tenure_stats.round(2)

# Afisam tabelul cu statisticile calculate
print("\nVechimea clientilor in functie de Churn:")
display(tenure_stats)
```

Comparăm distribuția vechimii clienților care au plecat și au rămas

```python

# Stabilim dimensiunea graficului
plt.figure(figsize=(9, 5))

# Construim un boxplot pentru a compara vechimea clientilor
# x reprezinta categoria Churn (No sau Yes)
# y reprezinta vechimea clientilor in luni
sns.boxplot(
    data=df_eda,
    x="Churn",
    y="tenure",
    order=["No", "Yes"],
    color="steelblue"
)

# Adaugam titlul graficului
plt.title("Distributia vechimii clientilor in functie de Churn")

# Denumim axele graficului
plt.xlabel("A renuntat clientul la servicii?")
plt.ylabel("Vechimea clientului (luni)")

# Afisam graficul
plt.show()
```

Intervalele de vechime. Grupăm clienții după vechime și calculăm rata abandonului

```python

# Definim intervalele de vechime exprimate in luni
# Folosim -1 pentru a include si clientii cu tenure = 0
intervale = [-1, 6, 12, 24, 48, 72]

# Stabilim denumirile categoriilor de vechime
etichete = ["0-6 luni", "7-12 luni", "13-24 luni",
            "25-48 luni", "49-72 luni"]

# Cream o coloana noua care incadreaza clientii in intervale
# pd.cut() transforma valorile numerice in categorii
df_eda["tenure_group"] = pd.cut(
    df_eda["tenure"],
    bins=intervale,
    labels=etichete
)

# Grupam clientii dupa intervalul de vechime
# count numara toti clientii din fiecare interval
# lambda numara doar clientii pentru care Churn este Yes
tenure_churn = df_eda.groupby(
    "tenure_group", observed=True
)["Churn"].agg(
    total_clienti="count",
    clienti_plecati=lambda x: (x == "Yes").sum()
)

# Calculam rata abandonului pentru fiecare interval
# Impartim numarul clientilor plecati la numarul total
# si inmultim cu 100 pentru a obtine procente
tenure_churn["rata_abandon"] = (
    tenure_churn["clienti_plecati"] /
    tenure_churn["total_clienti"] * 100
)

# Afisam rezultatele rotunjite la doua zecimale
print("Rata abandonului pe intervale de vechime:")
display(tenure_churn.round(2))
```

Reprezentăm rata abandonului pe intervale de vechime

```python

# Stabilim dimensiunea graficului
plt.figure(figsize=(10, 5))

# Construim graficul cu rata abandonului pe intervale
ax = sns.barplot(
    x=tenure_churn.index,
    y=tenure_churn["rata_abandon"].values,
    color="steelblue"
)

# Afisam procentele deasupra fiecarei bare
for i, valoare in enumerate(tenure_churn["rata_abandon"]):
    ax.text(
        i,
        valoare + 1,
        f"{valoare:.2f}%",
        ha="center"
    )

# Adaugam titlul graficului
plt.title("Rata abandonului in functie de vechimea clientilor")

# Denumim axele
plt.xlabel("Vechimea clientilor")
plt.ylabel("Rata abandonului (%)")

# Stabilim intervalul axei Y intre 0 si 100%
plt.ylim(0, 100)

# Afisam graficul
plt.show()
```

Analiza costurilor lunare. Calculăm statistici despre MonthlyCharges

```python

# Afisam statistici generale pentru costurile lunare
# describe() calculeaza media, mediana, minimul, maximul etc.
print("Statistici generale pentru MonthlyCharges:")
display(df_eda["MonthlyCharges"].describe().round(2))

# Grupam clientii in functie de variabila Churn
# Calculam statisticile costurilor lunare pentru fiecare categorie
monthly_stats = df_eda.groupby("Churn")["MonthlyCharges"].agg(
    ["count", "mean", "median", "std", "min", "max"]
)

# Rotunjim rezultatele la doua zecimale
monthly_stats = monthly_stats.round(2)

# Afisam tabelul cu statisticile calculate
print("\nCosturile lunare in functie de Churn:")
display(monthly_stats)
```

Grupăm costurile lunare și calculăm rata abandonului

```python

# Definim intervalele de cost lunar
# Folosim infinit pentru ca ultimul interval sa includa toate valorile mari
intervale_cost = [0, 40, 70, 90, float("inf")]

# Stabilim denumirile categoriilor de cost
etichete_cost = ["Sub 40", "40-69.99", "70-89.99", "90 sau mai mult"]

# Cream o coloana noua cu intervalul de cost al fiecarui client
# right=False inseamna ca limita din stanga este inclusa,
# iar limita din dreapta este exclusa
df_eda["monthly_group"] = pd.cut(
    df_eda["MonthlyCharges"],
    bins=intervale_cost,
    labels=etichete_cost,
    right=False
)

# Grupam clientii dupa intervalul costului lunar
# Numarul total reprezinta toti clientii din intervalul respectiv
# Numarul clientilor plecati este calculat pentru Churn = Yes
monthly_churn = df_eda.groupby(
    "monthly_group", observed=True
)["Churn"].agg(
    total_clienti="count",
    clienti_plecati=lambda x: (x == "Yes").sum()
)

# Calculam rata abandonului pentru fiecare interval
monthly_churn["rata_abandon"] = (
    monthly_churn["clienti_plecati"] /
    monthly_churn["total_clienti"] * 100
)

# Afisam tabelul cu rezultatele rotunjite
print("Rata abandonului pe intervale de cost lunar:")
display(monthly_churn.round(2))
```

Analizăm rata abandonului pentru OnlineSecurity și TechSupport.

```python

# Selectam doar clientii care beneficiaza de internet
# Excludem clientii pentru care InternetService este No
df_internet = df_eda[
    df_eda["InternetService"] != "No"
].copy()

# Definim serviciile suplimentare pe care dorim sa le analizam
servicii = ["OnlineSecurity", "TechSupport"]

# Cream un dictionar in care vom salva rezultatele
rezultate_servicii = {}

# Parcurgem fiecare serviciu din lista
for serviciu in servicii:

    # Grupam clientii dupa utilizarea serviciului (Yes sau No)
    # Calculam numarul total de clienti si numarul celor care au plecat
    statistici = df_internet.groupby(serviciu)["Churn"].agg(
        total_clienti="count",
        clienti_plecati=lambda x: (x == "Yes").sum()
    )

    # Calculam rata abandonului pentru fiecare categorie
    statistici["rata_abandon"] = (
        statistici["clienti_plecati"] /
        statistici["total_clienti"] * 100
    )

    # Pastram categoriile in ordinea No, Yes
    statistici = statistici.reindex(["No", "Yes"])

    # Salvam rezultatele pentru a le utiliza la grafice
    rezultate_servicii[serviciu] = statistici

    # Afisam tabelul cu rezultatele serviciului analizat
    print(f"\nAnaliza serviciului {serviciu}:")
    display(statistici.round(2))
```

Comparăm grafic ratele abandonului pentru cele două servicii suplimentare.

```python

# Cream doua grafice alaturate, unul pentru fiecare serviciu
# sharey=True foloseste aceeasi scara pe axa Y
fig, axes = plt.subplots(1, 2, figsize=(13, 5), sharey=True)

# Parcurgem cele doua servicii si graficele corespunzatoare
for ax, serviciu in zip(axes, servicii):

    # Extragem rata abandonului pentru serviciul curent
    rate = rezultate_servicii[serviciu]["rata_abandon"]

    # Construim graficul cu bare
    bare = ax.bar(
        rate.index,
        rate.values,
        color="steelblue"
    )

    # Afisam procentul deasupra fiecarei bare
    ax.bar_label(bare, fmt="%.2f%%", padding=3)

    # Adaugam titlul graficului
    ax.set_title(f"Rata abandonului - {serviciu}")

    # Denumim axele
    ax.set_xlabel("Utilizeaza serviciul?")
    ax.set_ylabel("Rata abandonului (%)")

    # Stabilim aceeasi scara pentru compararea graficelor
    ax.set_ylim(0, 100)

# Ajustam automat spatierea dintre grafice
plt.tight_layout()

# Afisam graficele
plt.show()
```

Analiza combinată. Calculăm rata abandonului după Contract și InternetService.

```python

# Grupam clientii simultan dupa tipul contractului
# si dupa tipul serviciului de internet
# Pentru fiecare combinatie calculam numarul total de clienti
# si numarul clientilor care au renuntat la servicii
contract_internet = df_eda.groupby(
    ["Contract", "InternetService"]
)["Churn"].agg(
    total_clienti="count",
    clienti_plecati=lambda x: (x == "Yes").sum()
)

# Calculam rata abandonului pentru fiecare combinatie
# Rata reprezinta procentul clientilor care au plecat
# din totalul clientilor cu acea combinatie de caracteristici
contract_internet["rata_abandon"] = (
    contract_internet["clienti_plecati"] /
    contract_internet["total_clienti"] * 100
)

# Afisam rezultatele rotunjite la doua zecimale
print("Rata abandonului in functie de Contract si InternetService:")
display(contract_internet.round(2))

# Transformam tabelul pentru reprezentarea grafica
# unstack() muta valorile InternetService pe coloane
# Contract ramane pe randuri
rata_tabel = contract_internet["rata_abandon"].unstack()

# Stabilim ordinea categoriilor pentru o afisare clara
rata_tabel = rata_tabel.reindex(
    index=["Month-to-month", "One year", "Two year"],
    columns=["DSL", "Fiber optic", "No"]
)

# Cream si un tabel cu numarul clientilor din fiecare combinatie
# Acesta ne ajuta sa interpretam corect ratele calculate
numar_tabel = contract_internet["total_clienti"].unstack()

# Pastram aceeasi ordine a categoriilor
numar_tabel = numar_tabel.reindex(
    index=rata_tabel.index,
    columns=rata_tabel.columns
)

# Afisam tabelul cu ratele abandonului
print("\nTabelul ratelor de abandon (%):")
display(rata_tabel.round(2))
```

Heatmap. Vizualizăm rata abandonului pentru fiecare combinație de contract și internet.

```python

# Stabilim dimensiunea graficului
plt.figure(figsize=(11, 6))

# Cream etichetele care vor aparea in fiecare celula a graficului
# Afisam atat rata abandonului, cat si numarul clientilor
etichete = rata_tabel.copy().astype(object)

# Parcurgem toate combinatiile de contract si serviciu
for contract in rata_tabel.index:
    for internet in rata_tabel.columns:

        # Extragem rata abandonului pentru combinatia curenta
        rata = rata_tabel.loc[contract, internet]

        # Extragem numarul clientilor pentru aceeasi combinatie
        numar = numar_tabel.loc[contract, internet]

        # Construim eticheta care contine procentul si numarul clientilor
        etichete.loc[contract, internet] = f"{rata:.1f}%\n(n={int(numar)})"

# Construim harta termica folosind Seaborn
# Culorile mai intense reprezinta rate mai ridicate de abandon
sns.heatmap(
    rata_tabel,
    annot=etichete,
    fmt="",
    cmap="Blues",
    vmin=0,
    vmax=100,
    linewidths=0.5,
    cbar_kws={"label": "Rata abandonului (%)"}
)

# Adaugam titlul graficului
plt.title("Rata abandonului: Contract si InternetService")

# Denumim axele graficului
plt.xlabel("Tipul serviciului de internet")
plt.ylabel("Tipul contractului")

# Ajustam spatierea pentru afisarea corecta
plt.tight_layout()

# Afisam graficul
plt.show()
```
